```python
# Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset

df = pd.read_csv('/content/Dataset .csv')

# Display First 5 Rows

df.head()
```

```python
# Distribution of Aggregate Ratings

plt.figure(figsize=(8,5))

sns.histplot(df['Aggregate rating'], bins=30, kde=True)

plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Count')

plt.show()
```

```python
# Top 10 Cities by Average Rating

top_cities = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

sns.barplot(x=top_cities.index, y=top_cities.values)

plt.title('Top 10 Cities by Average Rating')
plt.xlabel('City')
plt.ylabel('Average Rating')

plt.xticks(rotation=45)

plt.show()
```

```python
# Relationship Between Votes and Aggregate Rating

plt.figure(figsize=(8,5))

sns.scatterplot(x=df['Votes'], y=df['Aggregate rating'])

plt.title('Votes vs Aggregate Rating')
plt.xlabel('Votes')
plt.ylabel('Aggregate Rating')

plt.show()
```

```python
## Conclusion

- Multiple visualizations were created to analyze restaurant ratings and customer behavior.
- The distribution of aggregate ratings was visualized using a histogram.
- Top cities were compared based on their average restaurant ratings.
- The relationship between votes and aggregate ratings was analyzed using a scatter plot.
- Data visualization helped in identifying patterns and trends within the dataset.
```
