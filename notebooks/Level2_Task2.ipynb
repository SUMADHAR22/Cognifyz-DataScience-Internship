```python
# Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset

df = pd.read_csv('/content/Dataset .csv')

# Display First 5 Rows

df.head()
```

```python
# Most Common Price Range

price_range_count = df['Price range'].value_counts()

print(price_range_count)
```

```python
# Average Rating for Each Price Range

average_rating = df.groupby('Price range')['Aggregate rating'].mean()

print(average_rating)
```

```python
# Average Rating by Rating Color

color_rating = df.groupby('Rating color')['Aggregate rating'].mean().sort_values(ascending=False)

print(color_rating)
```

```python
# Visualization of Average Rating by Price Range

plt.figure(figsize=(8,5))

sns.barplot(x=average_rating.index, y=average_rating.values)

plt.title('Average Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')

plt.show()
```

```python
Conclusion

- Price range analysis was successfully performed on the restaurant dataset.
- The most common price range among restaurants was identified.
- Average aggregate ratings were calculated for each price category.
- Rating colors were analyzed to determine which category had the highest ratings.
- A visualization was created to better understand the relationship between price range and restaurant ratings.
```
